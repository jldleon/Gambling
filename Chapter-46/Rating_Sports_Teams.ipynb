# Least Square Ratings

Tenemos un conjunto de equipos en una competencia. Y queremos rankear esos equipos basandonos en sus resultados previos en la liga, basandonos en los margenes de victoria a lo largo de un calendario. 

La idea es asignar un rating objectivo en cuanta diferencia o que tan fuerte es un equipo con el resto de la liga.

Asumimos que el margen esperado de victoria en cualquier juego es directamente proporcional a la diferencia de los ratings de los rivales

$$ E[Y] = r_{home} - r_{away}$$

Siendo $Y$ una variable aleatoria que representa el margen de victoria entre dos equipos uno local y otro visitante.

Esta suposición implica que existe un ranking verdadero para cada equipo de tal manera que se cumple la exprsión: $ E[Y] = r_{home} - r_{away}$.


La meta de aisgnar ratings a los equipos, para explicar "mejor" los juegos completados, es equivalente a una multivarible regresion lineal observable.

Definición de una regresión lineal: 

$$ Y = \beta_0 + \beta_1 X_1 + ... + \beta_{m} X_m + \epsilon $$

Donde : 

- $Y$ es la variable dependiente o la variable de respuesta.
- $X_1$ , $X_2$ , ... , $X_m$ son la variables explicativas, independientes, o regresoras.
- $\beta_0 , \beta_1, ..., \beta_m$ son los parametros del modelo, miden la influencia que las variables explicativas tienen sobre el regrendio.

Usando esta definición podemos escribirla en nuestro terminos para este caso:

$$ y_i = r_0 + r_1 x_{i1} + ... + r_n x_{in} + \epsilon $$

Donde y:

- $y_i$ es la diferencia de puntos para el juego $i$
- Los $r$'s son los ratings que van a ser los parametros del modelo, con $r_0$ como la constante de home advantaje.
- Cada uno de los $n$ equipos es representado por es representado por una variable independiente predecible, donde puede tomar los valores 1, -1, 0. 1 para el equipo local, -1 para el equipo visitante, y 0 cuando ese equipo no participa en el juego $i$

Esto nos lleva a tener la ecuación de la predicción para un juego $i$:

$$ y_i = r_0 + r_{home} - r_{away} + \epsilon_{i} $$

Esto nos da la ecuación del libro:

Prediction Margin = (Home Edge) + (Home Team Rating) - (Away Team Rating)

Entonces el termino de error nos queda como:

$$ \epsilon_{i} = y_i - (r_0 + r_{home}  - r_{away} ) $$

Donde $r$ son los parametros de la regresion, que son obtenidos minimizando el error cuadrado: $ \sum \epsilon^{2}_{i} $


Si escribimos nuestro problema de forma matricial:

$$ \textbf{Y} = \textbf{X} \textbf{r} + \varepsilon $$

Si seguimos las matematicas de minimos cuadrados podemos resolver el sistema con:

$$ \textbf{r} = \textbf{X}^{+} \textbf{Y} $$

Donde $\textbf{X}^{+}$ es la pseudo inversa de $\textbf{X}$



```python
df_2016
```

```python
# Hay que construir la matriz X y el vector y

y = df_2016['Margin'].to_numpy()

y.shape

```

```python
# Hay que mapear los equipos a un directorio, hay que tener una lista por orden alfabetico

teams = sorted(teams)

teams
```

```python
teams_to_idx = {}

for i in range(len(teams)):
    teams_to_idx[teams[i]] = i + 1

teams_to_idx
```

```python
df_2016.reset_index(drop=True, inplace=True)

df_2016
```

```python
X = np.zeros((256, 33))

for i, game in df_2016.iterrows():
    # 1 en la primer columna para el r0
    X[i][0] = 1

    # 1 para el equipo local
    X[i][teams_to_idx[game['HomeTeam']]] = 1

    # -1 para el equipo visitante
    X[i][teams_to_idx[game['AwayTeam']]] = -1 
    
    
    
```

Para garantizar una solución unica se añade una nueva ecuación como restricción 

$$ r_0 + r_1 + ... + r_m = c = 0$$

```python
# Add the m + 1 ecuation r0 + r1 + ... + r3 = c = 0

X = np.vstack((X, np.ones(33)))

X.shape
```

```python
y = np.append(y, np.zeros(1))

y.shape
```

```python
X_pseudoinv = np.linalg.pinv(X)

r = X_pseudoinv.dot(y)
```

```python
HOME_EDGE = r[0]

print(f"Home Edge : {HOME_EDGE}")
```

```python
df_ratings = pd.DataFrame({ 'Team' : teams , 'Rating' : r[1:]})

df_ratings['Rank_LS'] = df_ratings['Rating'].rank(ascending=False) 

df_ratings
```

```python
team_to_rating = {}

for inx, team in df_ratings.iterrows():
    team_to_rating[team['Team']] = team['Rating']

print(team_to_rating)
```

```python
rating_map = df_ratings.set_index('Team')['Rating']

print(rating_map)

df_2016['Prediction'] = (
    HOME_EDGE
    + df_2016['HomeTeam'].map(rating_map)
    - df_2016['AwayTeam'].map(rating_map)
)
```

```python
df_2016['Squared Error'] = (df_2016['Margin'] - df_2016['Prediction'])**2

df_2016
```

# Evaluating Strength of Schedule

Basicamente para asignar un rating basado en el la fuerza del calendario, hay que sumar el rating anterior de todos sus rivales y dividirlo entre 16 (total de juegos)



```python
df_schedule_strength = df_ratings.copy()

df_schedule_strength
```

```python
print(team_to_rating)
```

```python
schedule_strenght = {team : 0 for team in team_to_rating}


for idx, game in df_2016.iterrows():
    schedule_strenght[game['HomeTeam']] += team_to_rating[game['AwayTeam']]
    schedule_strenght[game['AwayTeam']] += team_to_rating[game['HomeTeam']]


schedule_strenght = {team: total / 16 for team, total in schedule_strenght.items()}

df_ratings['schedule_strength'] = df_ratings['Team'].map(schedule_strenght)

df_ratings['Rank_SS'] = df_ratings['schedule_strength'].rank(ascending=False)

df_ratings
```
